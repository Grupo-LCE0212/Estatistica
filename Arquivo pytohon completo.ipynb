```python
import pandas as pd

file_path = 'dados.xlsx'

try:
    df = pd.read_excel(file_path)
    print("Dados importados com sucesso!")
    print("As primeiras 5 linhas do DataFrame são:")
    display(df.head(23))
except FileNotFoundError:
    print(f"Erro: O arquivo '{file_path}' não foi encontrado. Verifique o caminho do arquivo.")
except Exception as e:
    print(f"Ocorreu um erro ao ler o arquivo XLSX: {e}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Nome da coluna que o usuário pediu para plotar
column_name = "Qual dessas frutas você prefere para suco natural?"

# Verificar se a coluna existe no DataFrame
if column_name in df.columns:
    # Contar a frequência de cada fruta
    fruit_counts = df[column_name].value_counts()

    # Criar o gráfico de barras
    plt.figure(figsize=(10, 6))
    sns.barplot(x=fruit_counts.index, y=fruit_counts.values, palette='viridis')
    plt.title(f'Preferência de Frutas para Suco Natural ({column_name})')
    plt.xlabel('Frutas')
    plt.ylabel('Número de Respostas')
    plt.xticks(rotation=90, ha='right') # Rotacionar rótulos para melhor leitura
    plt.tight_layout()
    plt.show()
else:
    print(f"Erro: A coluna '{column_name}' não foi encontrada no DataFrame. Verifique o nome da coluna.")
```

### Gráficos para Variáveis Quantitativas Contínuas

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Assuming 'alturas', 'temperaturas', 'tempos_corrida' are already defined from previous cells
# If they are not defined, you might need to re-run the previous cell that generates them.

# Create some dummy data for demonstration if not already present
# These lines are for demonstration purposes if the kernel state does not retain previous variables.
if 'alturas' not in locals():
    alturas = np.random.uniform(1.50, 1.90, 100)
if 'temperaturas' not in locals():
    temperaturas = np.random.uniform(15.0, 30.0, 100)
if 'tempos_corrida' not in locals():
    tempos_corrida = np.random.uniform(10.0, 30.0, 100)


# Gráfico para Alturas
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.histplot(alturas, kde=True, bins=10)
plt.title('Distribuição das Alturas')
plt.xlabel('Altura (metros)')
plt.ylabel('Frequência')

# Gráfico para Temperaturas
plt.subplot(1, 3, 2)
sns.histplot(temperaturas, kde=True, bins=10)
plt.title('Distribuição das Temperaturas')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Frequência')

# Gráfico para Tempos de Corrida
plt.subplot(1, 3, 3)
sns.histplot(tempos_corrida, kde=True, bins=10)
plt.title('Distribuição dos Tempos de Corrida')
plt.xlabel('Tempo (segundos)')
plt.ylabel('Frequência')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Assuming 'numero_filhos', 'numero_carros', 'numero_erros' are already defined from previous cells
# If they are not defined, you might need to re-run the previous cell that generates them.

# Create some dummy data for demonstration if not already present
if 'numero_filhos' not in locals():
    numero_filhos = np.random.randint(0, 5, 100)
if 'numero_carros' not in locals():
    numero_carros = np.random.randint(0, 3, 100)
if 'numero_erros' not in locals():
    numero_erros = np.random.randint(0, 10, 100)


# Gráfico para Número de Filhos
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.countplot(x=numero_filhos, palette='viridis')
plt.title('Número de Filhos por Família')
plt.xlabel('Número de Filhos')
plt.ylabel('Contagem')

# Gráfico para Número de Carros
plt.subplot(1, 3, 2)
sns.countplot(x=numero_carros, palette='viridis')
plt.title('Número de Carros em Garagem')
plt.xlabel('Número de Carros')
plt.ylabel('Contagem')

# Gráfico para Número de Erros
plt.subplot(1, 3, 3)
sns.countplot(x=numero_erros, palette='viridis')
plt.title('Número de Erros em Texto')
plt.xlabel('Número de Erros')
plt.ylabel('Contagem')

plt.tight_layout()
plt.show()
```

### Gráficos para Variáveis Qualitativas Nominais

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Assuming 'cores_olhos', 'estado_civil', 'tipo_sanguineo' are already defined from previous cells
# If they are not defined, you might need to re-run the previous cell that generates them.

# Create some dummy data for demonstration if not already present
if 'cores_olhos' not in locals():
    cores_olhos = pd.Series(np.random.choice(['Castanho', 'Azul', 'Verde', 'Preto'], 100))
if 'estado_civil' not in locals():
    estado_civil = pd.Series(np.random.choice(['Solteiro', 'Casado', 'Divorciado', 'Viúvo'], 100))
if 'tipo_sanguineo' not in locals():
    tipo_sanguineo = pd.Series(np.random.choice(['A+', 'O-', 'B+', 'AB+'], 100))


# Gráfico para Cor dos Olhos
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.countplot(x=cores_olhos, palette='plasma')
plt.title('Cores dos Olhos')
plt.xlabel('Cor do Olho')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right')

# Gráfico para Estado Civil
plt.subplot(1, 3, 2)
sns.countplot(x=estado_civil, palette='plasma')
plt.title('Estado Civil')
plt.xlabel('Estado Civil')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right')

# Gráfico para Tipo Sanguíneo
plt.subplot(1, 3, 3)
sns.countplot(x=tipo_sanguineo, palette='plasma')
plt.title('Tipo Sanguíneo')
plt.xlabel('Tipo Sanguíneo')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

### Gráficos para Variáveis Qualitativas Ordinais

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Assuming 'escolaridade', 'satisfacao', 'classificacao_militar' are already defined from previous cells
# If they are not defined, you might need to re-run the previous cell that generates them.

# Create some dummy data for demonstration if not already present
if 'escolaridade' not in locals():
    escolaridade = pd.Series(np.random.choice(['Ensino Fundamental', 'Ensino Médio', 'Graduação', 'Pós-graduação'], 100),
                              dtype=pd.CategoricalDtype(categories=['Ensino Fundamental', 'Ensino Médio', 'Graduação', 'Pós-graduação'], ordered=True))
if 'satisfacao' not in locals():
    satisfacao = pd.Series(np.random.choice(['Insatisfeito', 'Neutro', 'Satisfeito', 'Muito Satisfeito'], 100),
                             dtype=pd.CategoricalDtype(categories=['Insatisfeito', 'Neutro', 'Satisfeito', 'Muito Satisfeito'], ordered=True))
if 'classificacao_militar' not in locals():
    classificacao_militar = pd.Series(np.random.choice(['Soldado', 'Cabo', 'Sargento', 'Tenente', 'Capitão'], 100),
                                   dtype=pd.CategoricalDtype(categories=['Soldado', 'Cabo', 'Sargento', 'Tenente', 'Capitão'], ordered=True))


# Gráfico para Nível de Escolaridade
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
sns.countplot(x=escolaridade, palette='mako', order=escolaridade.cat.categories)
plt.title('Nível de Escolaridade')
plt.xlabel('Escolaridade')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right')

# Gráfico para Nível de Satisfação
plt.subplot(1, 3, 2)
sns.countplot(x=satisfacao, palette='mako', order=satisfacao.cat.categories)
plt.title('Nível de Satisfação')
plt.xlabel('Satisfação')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right')

# Gráfico para Classificação Militar
plt.subplot(1, 3, 3)
sns.countplot(x=classificacao_militar, palette='mako', order=classificacao_militar.cat.categories)
plt.title('Classificação Militar')
plt.xlabel('Classificação')
plt.ylabel('Contagem')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
