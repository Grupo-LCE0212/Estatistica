#Semana 02 - Estatística

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file
file_path = '/content/dados_semana02.xlsx'
df = pd.read_excel(file_path)

# Display the first 5 rows of the DataFrame
print('Primeiras 5 linhas da base de dados:')
display(df.head())

# Display the column names
print('\nNomes das colunas:')
print(df.columns.tolist())
```

```python
# 1. Gráfico de pizza mostrando a participação de cada hortaliça na quantidade total vendida

# Calculate total quantity sold for each vegetable
hortalica_quantidade = df.groupby('Hortaliça')['Quantidade vendida (caixas)'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 8))
plt.pie(hortalica_quantidade, labels=hortalica_quantidade.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)
plt.title('Participação de Cada Hortaliça na Quantidade Total Vendida')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.legend(title='Hortaliças', loc='center left', bbox_to_anchor=(1, 0, 0.5, 1))
plt.tight_layout()
plt.show()
```

```python
# 2. Gráfico de barras mostrando a quantidade vendida por hortaliça

# Reuse hortaliça_quantidade calculated above

plt.figure(figsize=(12, 7))
sns.barplot(x=hortalica_quantidade.index, y=hortalica_quantidade.values, hue=hortalica_quantidade.index, palette='viridis', legend=False)
plt.title('Quantidade Vendida por Hortaliça')
plt.xlabel('Hortaliça')
plt.ylabel('Quantidade Total Vendida')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 3. Boxplot mostrando a distribuição dos preços por caixa

plt.figure(figsize=(10, 7))
sns.boxplot(y=df['Preço por caixa (R$)'], color='lightgreen') # Using color instead of palette for a single boxplot
plt.title('Distribuição dos Preços por Caixa')
plt.ylabel('Preço por Caixa')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 4. Histograma mostrando a distribuição dos preços por caixa

plt.figure(figsize=(10, 7))
sns.histplot(df['Preço por caixa (R$)'], kde=True, bins=15, color='skyblue', edgecolor='black')
plt.title('Distribuição dos Preços por Caixa')
plt.xlabel('Preço por Caixa')
plt.ylabel('Frequência')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
