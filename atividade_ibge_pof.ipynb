# Atividade em grupo – Estatística com dados de alimentos

## Tema
**Subamostra de consumo alimentar da Pesquisa de Orçamentos Familiares (POF) 2008–2009, por Unidade da Federação.**

Os dados abaixo foram retirados da **Tabela 1** da publicação do IBGE *Pesquisa de Orçamentos Familiares 2008–2009: Tabela de Medidas Referidas para os Alimentos Consumidos no Brasil*.

Foram selecionadas 20 Unidades da Federação.

A coluna **Região** foi acrescentada a partir da localização de cada UF. A coluna **Porte_subamostra** foi criada a partir do número de pessoas:
- Pequena: menos de 800 pessoas;
- Média: de 800 a 1.499 pessoas;
- Grande: 1.500 pessoas ou mais.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Construção da base de dados

```python
dados = {
    "UF": [
        "Rondônia", "Acre", "Amazonas", "Roraima", "Pará",
        "Amapá", "Tocantins", "Maranhão", "Piauí", "Ceará",
        "Rio Grande do Norte", "Paraíba", "Pernambuco", "Alagoas",
        "Sergipe", "Bahia", "Minas Gerais", "Espírito Santo",
        "Rio de Janeiro", "São Paulo"
    ],
    "Região": [
        "Norte", "Norte", "Norte", "Norte", "Norte",
        "Norte", "Norte", "Nordeste", "Nordeste", "Nordeste",
        "Nordeste", "Nordeste", "Nordeste", "Nordeste", "Nordeste",
        "Nordeste", "Sudeste", "Sudeste", "Sudeste", "Sudeste"
    ],
    "Setores_selecionados": [73,66,105,55,156,44,102,209,153,143,113,128,193,246,141,245,439,330,171,294],
    "Domicilios_POF": [907,863,1344,644,1894,689,1270,2562,2056,1861,1342,1628,2367,2712,1654,3050,5028,3489,1938,3623],
    "Domicilios_consumo": [230,235,402,149,470,161,333,599,548,441,330,342,582,649,449,850,1238,376,512,938],
    "Pessoas_subamostra": [569,608,1105,329,1266,499,898,1526,1551,1175,874,959,1499,1642,1150,2239,2960,841,1228,2273]
}

df = pd.DataFrame(dados)

df["Porte_subamostra"] = pd.cut(
    df["Pessoas_subamostra"],
    bins=[0, 799, 1499, float("inf")],
    labels=["Pequena", "Média", "Grande"],
    include_lowest=True
)

df
```

## 2. Tabela de frequência – variável nominal

```python
freq_regiao = df["Região"].value_counts().sort_index()
freq_rel_regiao = (df["Região"].value_counts(normalize=True).sort_index() * 100).round(1)

tabela_regiao = pd.DataFrame({
    "Frequência simples": freq_regiao,
    "Frequência relativa (%)": freq_rel_regiao
})
tabela_regiao
```

### Interpretação
A tabela mostra quantas observações da base pertencem às regiões Norte, Nordeste e Sudeste. Como a base contém apenas 20 UFs selecionadas, ela descreve somente essa amostra reduzida.


## 3. Tabela de frequência – variável ordinal

```python
ordem = ["Pequena", "Média", "Grande"]

freq_porte = df["Porte_subamostra"].value_counts().reindex(ordem)
freq_rel_porte = (df["Porte_subamostra"].value_counts(normalize=True).reindex(ordem) * 100).round(1)

tabela_porte = pd.DataFrame({
    "Frequência simples": freq_porte,
    "Frequência relativa (%)": freq_rel_porte
})
tabela_porte
```

### Interpretação
A variável ordinal permite comparar as UFs por tamanho da subamostra de pessoas em três níveis: pequena, média e grande.


## 4. Gráfico de barras

```python
plt.figure(figsize=(12, 6))
plt.bar(df["UF"], df["Pessoas_subamostra"])
plt.title("Número de pessoas na subamostra de consumo alimentar – POF 2008–2009")
plt.xlabel("Unidade da Federação")
plt.ylabel("Número de pessoas")
plt.xticks(rotation=70, ha="right")
plt.tight_layout()
plt.show()
```

### Interpretação
Entre as 20 UFs selecionadas, Minas Gerais apresenta o maior número de pessoas na subamostra, seguido por São Paulo e Bahia. Roraima apresenta o menor valor.


## 5. Histograma

```python
plt.figure(figsize=(9, 6))
plt.hist(df["Pessoas_subamostra"], bins=6, edgecolor="black")
plt.title("Distribuição do número de pessoas na subamostra de consumo alimentar")
plt.xlabel("Número de pessoas na subamostra")
plt.ylabel("Frequência de UFs")
plt.tight_layout()
plt.show()
```

### Interpretação
A maior parte das UFs selecionadas está concentrada nos valores intermediários de tamanho da subamostra, com algumas UFs apresentando valores mais elevados.


## 6. Gráfico de dispersão

```python
plt.figure(figsize=(9, 6))
plt.scatter(df["Domicilios_consumo"], df["Pessoas_subamostra"])
plt.title("Domicílios com consumo alimentar x pessoas na subamostra")
plt.xlabel("Domicílios com consumo alimentar")
plt.ylabel("Número de pessoas na subamostra")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretação
O gráfico mostra uma tendência positiva: UFs com maior número de domicílios incluídos na subamostra de consumo alimentar tendem também a apresentar maior número de pessoas participantes.


## 7. Mini-relatório

A análise utilizou dados reais da POF 2008–2009 do IBGE referentes à subamostra de consumo alimentar, considerando 20 Unidades da Federação.

As tabelas de frequência permitiram observar a composição da base por região e por porte da subamostra. O gráfico de barras mostrou diferenças no número de pessoas pesquisadas entre as UFs. O histograma mostrou a distribuição do tamanho dessas subamostras.

O gráfico de dispersão indicou uma associação positiva entre o número de domicílios com consumo alimentar e o número de pessoas na subamostra.
