### Estudo de Caso: Teor de Proteína em Alimento

Vamos identificar os elementos essenciais da distribuição:
- **Variável Aleatória de Interesse ($X$):** Teor de proteína em g/100 g de uma unidade de produto selecionada aleatoriamente.
- **Distribuição de Probabilidade:** Distribuição Normal.
- **Parâmetros da Distribuição:**
  - Média ($\mu$) = $12,5$ g/100 g
  - Desvio-padrão ($\sigma$) = $0,8$ g/100 g (Variância $\sigma^2 = 0,64$)

Notação matemática formal:
$$X \sim N(\mu = 12,5; \sigma^2 = 0,8^2)$$

#### 1. Distribuição de Probabilidade de $X$
Como especificado:
$$X \sim N(12,5; 0,64)$$
Onde a média é $\mu = 12,5$ e o desvio-padrão é $\sigma = 0,8$.

#### 2. Probabilidade de teor de proteína entre 11,5 e 13,5 g/100 g
A probabilidade de interesse é expressa como:
$$P(11,5 \le X \le 13,5)$$

Para calcular essa probabilidade:
$$P(11,5 \le X \le 13,5) = P\left(\frac{11,5 - 12,5}{0,8} \le Z \le \frac{13,5 - 12,5}{0,8}\right) = P(-1,25 \le Z \le 1,25)$$
Utilizando a função de distribuição acumulada (CDF) da distribuição Normal:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Definição dos parâmetros explicitamente
mu = 12.5
sigma = 0.8

x_inf = 11.5
x_sup = 13.5

# Cálculo utilizando a CDF da Normal
prob_1 = norm.cdf(x_sup, loc=mu, scale=sigma) - norm.cdf(x_inf, loc=mu, scale=sigma)
print(f"Expressão: P(11.5 <= X <= 13.5)")
print(f"Cálculo: CDF(13.5) - CDF(11.5)")
print(f"Resultado da Probabilidade: {prob_1:.4f} (ou {prob_1*100:.2f}%)")
```

#### 3. Representação Gráfica para o Intervalo $P(11,5 \le X \le 13,5)$

```python
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, loc=mu, scale=sigma)

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y, color='blue', linewidth=2, label='Curva de Densidade Normal $X \sim N(12.5, 0.8^2)$')

# Sombreamento da região de interesse
x_fill = np.linspace(x_inf, x_sup, 500)
y_fill = norm.pdf(x_fill, loc=mu, scale=sigma)
ax.fill_between(x_fill, y_fill, color='skyblue', alpha=0.5, label=f'Área: P(11.5 <= X <= 13.5) = {prob_1:.4f}')

# Linhas de referência
ax.axvline(mu, color='red', linestyle='--', label=f'Média (mu) = {mu}')
ax.axvline(x_inf, color='green', linestyle=':')
ax.axvline(x_sup, color='green', linestyle=':')

ax.set_title('Distribuição Normal do Teor de Proteína - Intervalo [11.5, 13.5]')
ax.set_xlabel('Teor de Proteína (g/100 g)')
ax.set_ylabel('Densidade de Probabilidade')
ax.legend(loc='upper right')
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

#### 4. Probabilidade de teor de proteína superior a 14 g/100 g
A probabilidade de interesse é expressa como:
$$P(X > 14)$$

Para calcular essa probabilidade:
$$P(X > 14) = 1 - P(X \le 14) = 1 - \text{CDF}(14)$$
Ou utilizando a função de sobrevivência (`sf`):

```python
limite_sup = 14
prob_2 = norm.sf(limite_sup, loc=mu, scale=sigma)
print(f"Expressão: P(X > 14)")
print(f"Cálculo: 1 - CDF(14)")
print(f"Resultado da Probabilidade: {prob_2:.4f} (ou {prob_2*100:.2f}%)")
```

#### 5. Representação Gráfica para a Probabilidade $P(X > 14)$

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y, color='blue', linewidth=2, label='Curva de Densidade Normal $X \sim N(12.5, 0.8^2)$')

# Sombreamento da região limite_sup
x_fill_2 = np.linspace(limite_sup, mu + 4*sigma, 500)
y_fill_2 = norm.pdf(x_fill_2, loc=mu, scale=sigma)
ax.fill_between(x_fill_2, y_fill_2, color='orange', alpha=0.5, label=f'Área: P(X > 14) = {prob_2:.4f}')

# Linhas de referência
ax.axvline(mu, color='red', linestyle='--', label=f'Média (mu) = {mu}')
ax.axvline(limite_sup, color='darkorange', linestyle=':')

ax.set_title('Distribuição Normal do Teor de Proteína - Região Superior a 14')
ax.set_xlabel('Teor de Proteína (g/100 g)')
ax.set_ylabel('Densidade de Probabilidade')
ax.legend(loc='upper right')
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

#### 6. Percentil 95 da Distribuição do Teor de Proteína
Queremos encontrar o ponto $x_{0,95}$ tal que:
$$P(X \le x_{0,95}) = 0,95$$
Calcularemos usando a função de ponto percentual (`ppf`), que é a inversa da CDF.

```python
percentil_95 = norm.ppf(0.95, loc=mu, scale=sigma)
print(f"Percentil 95 calculado: {percentil_95:.4f} g/100 g")
```

#### Gráfico de Destaque do Percentil 95

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y, color='blue', linewidth=2, label='Curva de Densidade Normal $X \sim N(12.5, 0.8^2)$')

# Sombreamento até o percentil 95 (95% da área acumulada à esquerda)
x_fill_pct = np.linspace(mu - 4*sigma, percentil_95, 500)
y_fill_pct = norm.pdf(x_fill_pct, loc=mu, scale=sigma)
ax.fill_between(x_fill_pct, y_fill_pct, color='lightgreen', alpha=0.5, label='Área de 95% Acumulada')

# Linha de destaque para o percentil 95
ax.axvline(percentil_95, color='darkgreen', linestyle='-', linewidth=2,
           label=f'Percentil 95 = {percentil_95:.2f} g/100g')
ax.axvline(mu, color='red', linestyle='--', label=f'Média (mu) = {mu}')

ax.set_title('Distribuição Normal - Percentil 95')
ax.set_xlabel('Teor de Proteína (g/100 g)')
ax.set_ylabel('Densidade de Probabilidade')
ax.legend(loc='upper left')
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

#### 7. Explicação no Contexto do Problema

O percentil 95 da distribuição do teor de proteína é igual a **13,82 g/100 g**.

No contexto deste produto alimentício, isso significa que **95% dos produtos analisados possuem um teor de proteína de até 13,82 g por cada 100 g de produto**. De maneira complementar, isso significa que apenas **5%** dos produtos desse tipo são muito ricos em proteína, ultrapassando o teor de **13,82 g/100 g -> 13,82% (m/m)**
