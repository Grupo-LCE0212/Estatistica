# Análise Exploratória de Dados da POF (Orçamentos Familiares para Alimentos) do IBGE

```python
import pandas as pd
import plotly.express as px

# Load the Excel file
file_path = '/content/banco_dados_IBGE_POF.xlsx'
df = pd.read_excel(file_path)

print("Dados da amostra da POF (Pesquisa de Orçamentos Familiares para Alimentos) carregados com sucesso. Aqui estão as 5 primeiras linhas:")
display(df.head())

print("\nE aqui estão as colunas disponíveis no conjunto de dados, que descrevem a amostra da pesquisa:")
print(df.columns.tolist())
```

## Tabelas de Frequência e Interpretação no Contexto de Orçamentos Familiares para Alimentos

### Variável Nominal: `Região`

```python
# Frequências para a variável nominal 'Região'
freq_regiao = df['Região'].value_counts().reset_index()
freq_regiao.columns = ['Região', 'Frequência Simples']
freq_regiao['Frequência Relativa (%)'] = (freq_regiao['Frequência Simples'] / freq_regiao['Frequência Simples'].sum() * 100).round(2)

display(freq_regiao)
```

#### Interpretação da Frequência de `Região` no Contexto da POF de Alimentos:

Esta tabela nos mostra como os estados (que são as unidades de amostragem neste conjunto de dados) estão distribuídos pelas grandes regiões do Brasil. Para a pesquisa de orçamentos familiares com foco em alimentos, entender essa distribuição é chave: regiões como o **Nordeste**, que aparece com a maior frequência, indicam onde a POF teve uma cobertura mais ampla em termos de unidades amostrais. Isso sugere que teremos mais informações para entender os hábitos alimentares e os gastos com comida nessas áreas, tornando nossas análises sobre o consumo alimentar regional mais robustas e detalhadas.

### Variável Discreta: `Domicilios_POF`

```python
# Frequências para a variável discreta 'Domicilios_POF'
# Como 'Domicilios_POF' pode ter muitos valores únicos, vamos mostrar os 10 valores mais comuns e um sumário estatístico.

print("Top 10 valores mais comuns para 'Domicilios_POF' (número de domicílios amostrados para a POF de alimentos):")
freq_domicilios_top10 = df['Domicilios_POF'].value_counts().nlargest(10).reset_index()
freq_domicilios_top10.columns = ['Domicilios_POF', 'Frequência Simples']
freq_domicilios_top10['Frequência Relativa (%)'] = (freq_domicilios_top10['Frequência Simples'] / df['Domicilios_POF'].count() * 100).round(2)

display(freq_domicilios_top10)

print("\nSumário estatístico para 'Domicilios_POF' (número de domicílios amostrados):")
display(df['Domicilios_POF'].describe())
```

#### Interpretação da Frequência de `Domicilios_POF` no Contexto da POF de Alimentos:

A variável `Domicilios_POF` representa a quantidade de casas (domicílios) que foram incluídas na amostra para coletar dados sobre os gastos e o consumo de alimentos das famílias. Analisando os valores mais comuns e o resumo estatístico, vemos que o número de domicílios amostrados varia bastante em cada 'registro' (que pode ser um estado ou uma sub-região). Com uma média de aproximadamente 1400 domicílios por registro, e um desvio padrão considerável, isso nos diz que a 'força' da pesquisa de consumo alimentar pode ser bem diferente de um lugar para outro. Ou seja, a quantidade de informações detalhadas sobre a alimentação pode ser maior em alguns locais do que em outros.

## Gráficos com Plotly Express: Visualizando os Orçamentos Familiares para Alimentos

### 1. Gráfico de Barras: Quanto se Amostra de Domicílios por Região (para entender o consumo de alimentos)

```python
import matplotlib.pyplot as plt

# Agregando os dados para o gráfico de barras
df_regiao_domicilios = df.groupby('Região')['Domicilios_POF'].sum().reset_index()

plt.figure(figsize=(10, 6))
plt.bar(df_regiao_domicilios['Região'], df_regiao_domicilios['Domicilios_POF'], color=['skyblue', 'lightcoral', 'lightgreen', 'gold', 'plum'])
plt.title('Total de Domicílios Amostrados para POF de Alimentos por Região')
plt.xlabel('Região do Brasil')
plt.ylabel('Total de Domicílios POF (Base para Dados de Alimentos)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

#### Interpretação do Gráfico de Barras:

Este gráfico nos permite ver rapidamente quais regiões têm a maior 'base' de domicílios para a pesquisa de orçamentos familiares focada em alimentos. Claramente, **Sudeste** e **Nordeste** se destacam, com um número muito maior de domicílios amostrados. Isso é crucial: significa que os dados de consumo alimentar dessas regiões tendem a ser mais detalhados e representativos, o que nos ajuda a tirar conclusões mais precisas sobre como as famílias gastam com comida e o que consomem nesses locais. Já em regiões com menos domicílios, a análise pode ser um pouco menos detalhada.

### 2. Histograma: A Variedade de Pessoas Amostradas para Observar o Consumo de Alimentos

```python
import matplotlib.pyplot as plt

# Histograma
plt.figure(figsize=(10, 6))
plt.hist(df['Pessoas_subamostra'], bins=20, edgecolor='black', alpha=0.7, color='teal')
plt.title('Distribuição de Pessoas na Subamostra da POF de Alimentos')
plt.xlabel('Número de Pessoas na Subamostra (Base para Dados de Alimentos)')
plt.ylabel('Frequência')
plt.tight_layout()
plt.show()
```

#### Interpretação do Histograma:

O histograma de `Pessoas_subamostra` mostra a distribuição de quantas pessoas foram incluídas em cada 'pedaço' da amostra da POF. Para o estudo de alimentos, isso é importante porque cada uma dessas pessoas contribui com informações sobre seu consumo. Vemos que a maioria dos registros tem um número relativamente baixo de pessoas na subamostra. Isso nos ajuda a entender a diversidade e o tamanho dos grupos cujos hábitos alimentares e despesas com comida estão sendo capturados pela pesquisa.

### 3. Gráfico de Dispersão: A Relação Entre Domicílios e Pessoas na Amostra (para entender o impacto nos Orçamentos de Alimentos)

```python
import matplotlib.pyplot as plt

# Gráfico de dispersão
plt.figure(figsize=(10, 6))
plt.scatter(df['Domicilios_POF'], df['Pessoas_subamostra'], alpha=0.7, color='purple')
plt.title('Relação entre Domicílios e Pessoas na Amostra da POF de Alimentos')
plt.xlabel('Número de Domicílios Amostrados (para Orçamentos de Alimentos)')
plt.ylabel('Número de Pessoas na Subamostra (para Consumo de Alimentos)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

#### Interpretação do Gráfico de Dispersão:

Este gráfico nos ajuda a visualizar a conexão entre o número de casas pesquisadas (`Domicilios_POF`) e o total de pessoas incluídas na subamostra (`Pessoas_subamostra`) dentro do contexto de coletar dados sobre o consumo e os gastos com alimentos. É esperado, e o gráfico confirma, uma **forte relação positiva**: quanto mais domicílios são pesquisados em uma área, mais pessoas naturalmente estarão lá para fornecer informações sobre seus orçamentos familiares para alimentos. Esta visualização garante que a estrutura da amostragem faz sentido e que mais domicílios realmente significam mais informações individuais sobre o que se come e quanto se gasta com comida.
