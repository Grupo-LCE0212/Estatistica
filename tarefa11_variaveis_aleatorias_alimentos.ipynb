A Probabilidade permite estudar fenômenos cujo resultado não pode ser conhecido com certeza antes de sua ocorrência. Nas Ciências dos Alimentos, esses fenômenos podem estar relacionados ao controle de qualidade, produção, armazenamento, segurança dos alimentos e avaliação de produtos.

### Variável Aleatória

Uma **variável aleatória** é uma variável que associa um valor numérico aos possíveis resultados de um experimento aleatório. Seu valor não é conhecido antes da realização do experimento, embora seja possível determinar quais valores ela pode assumir.

**Exemplo aplicado às Ciências dos Alimentos:**

Considere a inspeção de **10 embalagens de iogurte** para verificar a presença de defeitos. Podemos definir:

**X = número de embalagens defeituosas entre as 10 analisadas.**

Nesse caso, a variável aleatória X pode assumir os valores:

**X = {0, 1, 2, 3, ..., 10}**

O valor de X só será conhecido após a realização da inspeção. Por exemplo, se forem encontradas 2 embalagens defeituosas, teremos **X = 2**.

Essa variável também é classificada como **discreta**, pois representa uma contagem e pode assumir apenas valores inteiros.

### Variáveis Aleatórias Discretas e Contínuas

As variáveis aleatórias podem ser classificadas em **discretas** ou **contínuas**, de acordo com os valores que podem assumir.

#### Variáveis aleatórias discretas

Uma variável aleatória **discreta** assume valores que podem ser contados, geralmente números inteiros.

**Exemplos aplicados às Ciências dos Alimentos:**

1. **Número de embalagens reprovadas em um lote:** é discreta porque representa uma contagem e só pode assumir valores inteiros, como 0, 1, 2, 3 etc.

2. **Número de frutas danificadas em uma caixa:** é discreta porque a quantidade de frutas pode ser contada individualmente.

3. **Número de reclamações de consumidores recebidas por dia:** é discreta porque cada reclamação é uma ocorrência que pode ser contada.

#### Variáveis aleatórias contínuas

Uma variável aleatória **contínua** pode assumir qualquer valor dentro de um determinado intervalo, incluindo valores decimais, dependendo da precisão da medição.

**Exemplos aplicados às Ciências dos Alimentos:**

1. **Massa de um produto:** é contínua porque pode assumir diferentes valores dentro de um intervalo, como 499,5 g, 500,0 g ou 500,27 g.

2. **Temperatura de armazenamento:** é contínua porque pode assumir qualquer valor dentro de uma faixa de temperatura, como 4,0 °C, 4,5 °C ou 4,73 °C.

3. **pH de uma amostra de alimento:** é contínua porque é obtido por medição e pode assumir diferentes valores decimais dentro de uma faixa.

De forma resumida, **variáveis discretas estão relacionadas principalmente a contagens**, enquanto **variáveis contínuas estão relacionadas a medições**.

### Modelo de Bernoulli

O **modelo de Bernoulli** é utilizado para representar um experimento aleatório que possui apenas **dois resultados possíveis**, geralmente representados por sucesso (1) e fracasso (0).

Neste exemplo, será analisada a **selagem de uma embalagem de iogurte**:

- **Sucesso (1):** embalagem aprovada quanto à selagem;
- **Fracasso (0):** embalagem reprovada quanto à selagem.

Para fins didáticos, será considerada uma probabilidade simulada de **90% de aprovação** da embalagem. Portanto:

**P(sucesso) = 0,90**

**P(fracasso) = 0,10**

Os valores utilizados são simulados e não representam dados reais de uma indústria.

```python
# Modelo de Bernoulli aplicado à inspeção de uma embalagem de iogurte

import random

# Probabilidade simulada de a embalagem ser aprovada
probabilidade_aprovacao = 0.90

# Gera um número aleatório entre 0 e 1
resultado_aleatorio = random.random()

# Verifica o resultado da inspeção
if resultado_aleatorio < probabilidade_aprovacao:
    resultado = 1
    print("Resultado: Sucesso (1)")
    print("A embalagem foi aprovada quanto à selagem.")
else:
    resultado = 0
    print("Resultado: Fracasso (0)")
    print("A embalagem foi reprovada quanto à selagem.")
```

```python
import matplotlib.pyplot as plt

resultados = ["Reprovada (0)", "Aprovada (1)"]
probabilidades = [0.10, 0.90]

plt.bar(resultados, probabilidades)

plt.title("Modelo de Bernoulli - Selagem de Embalagem")
plt.xlabel("Resultado da inspeção")
plt.ylabel("Probabilidade")

plt.ylim(0, 1)

plt.show()
```

### Interpretação

O resultado gerado representa a inspeção simulada de **uma única embalagem**. O valor **1 representa sucesso (embalagem aprovada)** e o valor **0 representa fracasso (embalagem reprovada)**.

Como existem apenas dois resultados possíveis para cada embalagem analisada, essa situação pode ser representada pelo **modelo de Bernoulli**.

### Modelo Binomial

O **modelo Binomial** é utilizado para representar o número de sucessos obtidos em uma quantidade fixa de observações ou ensaios.

Para que esse modelo seja aplicado, devem existir algumas características:

- Um **número fixo de observações**;
- Cada observação deve apresentar apenas **dois resultados possíveis**, sucesso ou fracasso;
- A **probabilidade de sucesso deve permanecer constante** em cada observação;
- As observações são consideradas **independentes**.

Neste exemplo, serão analisadas **20 embalagens de iogurte** quanto à qualidade da selagem.

Consideramos:

- **n = 20:** número de embalagens analisadas;
- **Sucesso (1):** embalagem aprovada quanto à selagem;
- **Fracasso (0):** embalagem reprovada quanto à selagem;
- **p = 0,90:** probabilidade simulada de uma embalagem ser aprovada.

A variável aleatória **X** representa o **número de embalagens aprovadas entre as 20 analisadas**.

Assim, X pode assumir valores de **0 a 20**.

Os valores utilizados são simulados para fins didáticos e não representam dados reais de uma indústria.

```python
# Modelo Binomial aplicado à inspeção de embalagens de iogurte

import random

# Parâmetros do experimento
numero_embalagens = 20
probabilidade_aprovacao = 0.90

# Contador de embalagens aprovadas
embalagens_aprovadas = 0

# Simulação da inspeção das 20 embalagens
for i in range(numero_embalagens):

    resultado_aleatorio = random.random()

    if resultado_aleatorio < probabilidade_aprovacao:
        embalagens_aprovadas += 1

# Resultado
print("Número de embalagens analisadas:", numero_embalagens)
print("Número de embalagens aprovadas:", embalagens_aprovadas)
print("Número de embalagens reprovadas:",
      numero_embalagens - embalagens_aprovadas)
```

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parâmetros
n = 20
p = 0.90

# Valores possíveis: de 0 a 20 embalagens aprovadas
x = range(n + 1)

# Probabilidade de cada resultado
probabilidades = binom.pmf(x, n, p)

plt.bar(x, probabilidades)

plt.title("Distribuição Binomial - Embalagens Aprovadas")
plt.xlabel("Número de embalagens aprovadas")
plt.ylabel("Probabilidade")

plt.xticks(x)

plt.show()
```

### Interpretação

O código simula a inspeção de **20 embalagens de iogurte**. Cada embalagem pode apresentar apenas dois resultados: **aprovada ou reprovada**.

Ao final, é contado o número total de embalagens aprovadas. Por exemplo, se o resultado obtido for **18 embalagens aprovadas**, significa que ocorreram **18 sucessos em 20 observações**.

Diferentemente do modelo de Bernoulli, que representa **uma única observação**, o modelo Binomial representa o **número de sucessos obtidos em um número fixo de observações**.

### Modelo de Poisson

O **modelo de Poisson** é utilizado para representar o número de vezes que determinado evento ocorre dentro de um intervalo definido, que pode ser de tempo, espaço, área, volume ou unidade de produção.

Neste exemplo, será considerado o **número de falhas de selagem registradas por hora em uma linha de envase de iogurte**.

Para fins didáticos, consideramos que a linha apresenta, em média, **3 falhas de selagem por hora**.

Assim:

- **X = número de falhas de selagem observadas em uma hora;**
- **λ (lambda) = 3 falhas por hora**, representando a taxa média de ocorrências;
- X pode assumir os valores **0, 1, 2, 3, ...**

Os valores utilizados neste exemplo são simulados para fins didáticos e não representam dados reais de uma indústria.

```python
# Modelo de Poisson aplicado a uma linha de envase de iogurte

import math

# Número médio de falhas por hora
media_falhas = 3

# Número de falhas que queremos analisar
numero_falhas = 2

# Fórmula da distribuição de Poisson
probabilidade = (
    math.exp(-media_falhas)
    * (media_falhas ** numero_falhas)
    / math.factorial(numero_falhas)
)

print("Média de falhas por hora:", media_falhas)
print("Número de falhas analisado:", numero_falhas)
print(f"Probabilidade: {probabilidade:.4f}")
print(f"Probabilidade em porcentagem: {probabilidade * 100:.2f}%")
```

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Média simulada de falhas por hora
media_falhas = 3

# Quantidades de falhas que serão apresentadas
x = range(0, 11)

# Probabilidade de cada quantidade de falhas
probabilidades = poisson.pmf(x, mu=media_falhas)

plt.bar(x, probabilidades)

plt.title("Distribuição de Poisson - Falhas de Selagem")
plt.xlabel("Número de falhas por hora")
plt.ylabel("Probabilidade")

plt.xticks(x)

plt.show()
```

### Interpretação

Considerando uma média hipotética de **3 falhas de selagem por hora**, o modelo de Poisson permite calcular a probabilidade de observar determinada quantidade de falhas durante uma hora de produção.

Neste exemplo, a probabilidade de ocorrerem **exatamente 2 falhas em uma hora é de aproximadamente 22,40%**.

O modelo de Poisson é adequado para esse exemplo porque estamos realizando uma **contagem de ocorrências dentro de um intervalo definido de tempo**, diferentemente do modelo Binomial, no qual existe um número fixo de unidades analisadas.

### Comparação dos Modelos

Os modelos de **Bernoulli, Binomial e Poisson** são utilizados para representar diferentes tipos de fenômenos aleatórios. A tabela abaixo apresenta as principais diferenças entre eles e suas aplicações no contexto das Ciências dos Alimentos.

### Interpretação

Os três modelos representam situações diferentes. O modelo de **Bernoulli** é utilizado para uma única observação com dois resultados possíveis. O modelo **Binomial** amplia essa ideia para um número fixo de observações e permite contar quantos sucessos ocorreram. Já o modelo de **Poisson** é utilizado para contar quantas vezes determinado evento ocorre dentro de um intervalo definido, como uma hora de produção.

No contexto apresentado, Bernoulli representa a avaliação de **uma embalagem**, Binomial representa a avaliação de **20 embalagens** e Poisson representa a quantidade de **falhas de selagem registradas durante uma hora de produção**.
