### Análise do Problema: Embalagens Abaixo do Peso

Neste problema, estamos lidando com um cenário de controle de qualidade onde a probabilidade de uma embalagem estar abaixo do peso é fixa, temos um número definido de embalagens inspecionadas, e cada inspeção é independente. Isso aponta para a utilização da **Distribuição Binomial**.

#### 1. Quais valores a variável aleatória $X$ pode assumir?

A variável aleatória $X$ representa o número de embalagens abaixo do peso entre as 20 selecionadas. Como $X$ é uma contagem de sucessos em um número fixo de tentativas, seus valores possíveis são números inteiros de 0 (nenhuma embalagem abaixo do peso) a 20 (todas as embalagens abaixo do peso).

**Valores possíveis:** $X \in \{0, 1, 2, \dots, 20\}$.

O valor 0 representa a situação em que nenhuma das embalagens está abaixo do peso, enquanto o valor 20 representa a situação em que todas as embalagens avaliadas estão abaixo do peso. Os valores intermediários representam a quantidade correspondente de embalagens abaixo da especificação.
Como \(X\) é uma variável de contagem, ela assume apenas valores inteiros, não sendo possível, por exemplo, obter \(X=1,5\) embalagem.

#### 2. Qual distribuição de probabilidade pode ser utilizada para representar $X$? Identifique seus parâmetros e justifique sua escolha.

**Distribuição:** Distribuição Binomial.

**Parâmetros:**
*   $n$ (número de tentativas/embalagens inspecionadas) = 20
*   $p$ (probabilidade de sucesso/embalagem estar abaixo do peso) = 0.04

**Justificativa:**
Esta escolha é justificada pelas seguintes características do experimento:
1.  **Número fixo de tentativas:** São inspecionadas 20 embalagens ($n = 20$).
2.  **Duas categorias de resultado:** Cada embalagem pode estar "abaixo do peso" (sucesso) ou "não abaixo do peso" (fracasso).
3.  **Independência das tentativas:** A condição de uma embalagem não afeta a das outras.
4.  **Probabilidade de sucesso constante:** A probabilidade de uma embalagem estar abaixo do peso é a mesma para todas ($p = 0.04$).

Assim, $X \sim B(20, 0.04)$.

```python
from scipy.stats import binom

# Definindo os parâmetros da distribuição binomial
n = 20  # Número de tentativas (embalagens)
p = 0.04 # Probabilidade de sucesso (embalagem abaixo do peso)

print(f"Parâmetros da Distribuição Binomial: n = {n}, p = {p}")
```

A distribuição Binomial é adequada porque o experimento possui um número fixo de observações, correspondente às 20 embalagens selecionadas. Para cada embalagem existem dois resultados possíveis em relação ao evento estudado: ela pode estar abaixo do peso especificado ou não estar abaixo do peso especificado.
Além disso, o enunciado estabelece que as embalagens são independentes e produzidas sob condições semelhantes, permitindo considerar constante a probabilidade de \(4\%\) para cada embalagem.
Assim, \(X\) representa a contagem do número de vezes que o evento “embalagem abaixo do peso” ocorre nas 20 avaliações, característica típica de uma distribuição Binomial.
Resposta resumida: a distribuição é Binomial porque estamos contando quantas embalagens apresentam determinada característica em um número fixo de 20 observações independentes, cada uma com probabilidade \(p=0,04\) de apresentar essa característica.

#### 3. Qual é a probabilidade de nenhuma das 20 embalagens estar abaixo do peso especificado?

Queremos calcular $P(X = 0)$, ou seja, a probabilidade de que zero embalagens estejam abaixo do peso. Usando a função de massa de probabilidade (PMF) da distribuição binomial:

$P(X=k) = C(n, k) * p^k * (1-p)^{(n-k)}$

Para $k=0$:

$P(X=0) = C(20, 0) * (0.04)^0 * (0.96)^{20}$

```python
# Probabilidade de X = 0 (nenhuma embalagem abaixo do peso)
prob_x_0 = binom.pmf(0, n, p)
print(f"P(X = 0) = {prob_x_0:.4f}")
print(f"Interpretação: Há uma probabilidade de {prob_x_0:.2%} de que nenhuma das 20 embalagens esteja abaixo do peso especificado.")
```

Cada embalagem possui probabilidade de \(0,96\) de não estar abaixo do peso, pois \(1-0,04=0,96\). Como as 20 embalagens são independentes, a probabilidade de todas as 20 não estarem abaixo do peso é \((0,96)^{20}\). Portanto, há aproximadamente 44,20% de chance de nenhuma embalagem estar abaixo do peso especificado.     Distribuições Discretas

#### 4. Qual é a probabilidade de exatamente duas embalagens estarem abaixo do peso?

Queremos calcular $P(X = 2)$, ou seja, a probabilidade de que exatamente duas embalagens estejam abaixo do peso.

Para $k=2$:

$P(X=2) = C(20, 2) * (0.04)^2 * (0.96)^{18}$

```python
# Probabilidade de X = 2 (exatamente duas embalagens abaixo do peso)
prob_x_2 = binom.pmf(2, n, p)
print(f"P(X = 2) = {prob_x_2:.4f}")
print(f"Interpretação: Há uma probabilidade de {prob_x_2:.2%} de que exatamente duas das 20 embalagens estejam abaixo do peso.")
```

Utilizamos \(X=2\) porque queremos que exatamente duas das 20 embalagens estejam abaixo do peso. O termo \((0,04)^2\) representa a ocorrência de duas embalagens abaixo do peso, enquanto \((0,96)^{18}\) representa as outras 18 embalagens que não estão abaixo do peso. O termo \(\binom{20}{2}=190\) considera as diferentes maneiras pelas quais essas duas embalagens podem aparecer entre as 20 avaliadas.
Interpretação: há aproximadamente 14,60% de probabilidade de exatamente duas das 20 embalagens avaliadas estarem abaixo do peso especificado.

#### 5. Qual é a probabilidade de pelo menos uma embalagem estar abaixo do peso? Existe uma forma mais simples de calcular essa probabilidade utilizando o evento complementar?

Queremos calcular $P(X \ge 1)$, ou seja, a probabilidade de que uma ou mais embalagens estejam abaixo do peso.

Sim, existe uma forma mais simples utilizando o **evento complementar**.

$P(X \ge 1) = 1 - P(X < 1) = 1 - P(X = 0)$

Como já calculamos $P(X=0)$ na questão 3, podemos usar esse valor.

```python
# Probabilidade de X >= 1 (pelo menos uma embalagem abaixo do peso)
# Usando o evento complementar: P(X >= 1) = 1 - P(X = 0)
prob_x_ge_1 = 1 - prob_x_0
print(f"P(X >= 1) = {prob_x_ge_1:.4f}")
print(f"Interpretação: Há uma probabilidade de {prob_x_ge_1:.2%} de que pelo menos uma das 20 embalagens esteja abaixo do peso.")
```

Utilizamos o evento complementar porque seria necessário somar as probabilidades de \(X=1,2,3,\ldots,20\). É mais simples calcular a probabilidade de nenhuma embalagem estar abaixo do peso e subtrair esse resultado de 1.
Interpretação: existe aproximadamente 55,80% de probabilidade de que pelo menos uma das 20 embalagens avaliadas esteja abaixo do peso especificado.

#### 6. Determine $E(X)$. Quantas embalagens abaixo do peso são esperadas, em média, em conjuntos de 20 embalagens avaliadas sob essas condições? Interprete os resultados obtidos no contexto do controle de qualidade.

Para uma distribuição binomial, o valor esperado $E(X)$ é dado por:

$E(X) = n * p$

```python
# Valor esperado E(X)
expected_value_x = n * p
print(f"E(X) = {expected_value_x:.2f}")
print(f"Interpretação: Em média, espera-se que {expected_value_x:.2f} embalagens estejam abaixo do peso em cada lote de 20 embalagens inspecionadas.")
print("Contexto de Controle de Qualidade: Este valor esperado de 0.8 significa que, embora a probabilidade de uma única embalagem estar abaixo do peso seja pequena (4%), ao inspecionar 20 embalagens, é razoável esperar encontrar quase uma embalagem (0.8) fora do padrão em média. Isso é uma métrica importante para monitorar a qualidade do processo de produção. Se o número de embalagens abaixo do peso observado consistentemente exceder este valor, pode ser um indicativo de que o processo de produção está se desviando e necessita de ajustes ou investigação.")
```

Justificativa: o valor esperado de uma distribuição Binomial representa o número médio de ocorrências do evento em um conjunto de \(n\) tentativas. Neste caso, cada embalagem possui 4% de probabilidade de estar abaixo do peso e são avaliadas 20 embalagens.     Distribuições Discretas
Interpretação: espera-se encontrar, em média, 0,8 embalagem abaixo do peso a cada grupo de 20 embalagens avaliadas. Isso não significa que será encontrada 0,8 embalagem em uma inspeção específica, pois \(X\) só assume valores inteiros. O valor 0,8 representa uma média esperada ao realizar muitas inspeções de grupos de 20 embalagens.

### Gráfico da Distribuição de Probabilidade de $X$

Para visualizar a distribuição de probabilidade, podemos plotar um gráfico de barras que mostra $P(X=k)$ para cada valor possível de $k$ (número de embalagens abaixo do peso).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gerar os valores possíveis para X (de 0 a n)
k_values = list(range(n + 1))

# Calcular a PMF para cada valor de k
pmf_values = [binom.pmf(k, n, p) for k in k_values]

# Criar o gráfico de barras
plt.figure(figsize=(10, 6))
sns.barplot(x=k_values, y=pmf_values, palette='viridis', hue=k_values, legend=False)

plt.title('Distribuição de Probabilidade do Número de Embalagens Abaixo do Peso (X)')
plt.xlabel('Número de Embalagens Abaixo do Peso (k)')
plt.ylabel('Probabilidade P(X=k)')
plt.xticks(rotation=45) # Rotaciona os rótulos do eixo X se houver muitos
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
